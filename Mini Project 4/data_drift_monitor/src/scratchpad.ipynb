Loading the data and doing EDA as needed

```python
import sklearn.datasets
import pandas as pd
import numpy as np

data = sklearn.datasets.fetch_california_housing(as_frame=True)
df = data.frame
```

```python
df.head()
```

Saving the data in two CSV files 

```python
reference_df = df.iloc[:15000]
live_batch_df = df.iloc[15000:]

print("Saving reference and live batch data to CSV files...")
reference_df.to_csv("../data/reference_data.csv", index=False)
live_batch_df.to_csv("../data/live_batch_data.csv", index=False)
```

Making a Drifted data by doing Covariate Shift and Sensor/Data Engineering Failure

```python
live_batch_df = pd.read_csv("../data/live_batch_data.csv")

print("Modifying live batch data to simulate drift...")
live_batch_df['MedInc'] = live_batch_df['MedInc']*1.4

print("Introducing random noise in 'HouseAge' column to simulate drift...")
corrupted_indices = live_batch_df.sample(frac=0.3, random_state=42).index
live_batch_df.loc[corrupted_indices, 'HouseAge'] = 0
print("Saving modified live batch data to CSV file...")
live_batch_df.to_csv("../data/live_batch_data.csv", index=False)
```
